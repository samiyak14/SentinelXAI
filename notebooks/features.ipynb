```python
import pandas as pd
from pathlib import Path

csv_path = Path(
    r"C:\SentinelXAI\data\raw\MachineLearningCSV\MachineLearningCVE"
) / "Friday-WorkingHours-Morning.pcap_ISCX.csv"

# Read only the header
columns = pd.read_csv(csv_path, nrows=0).columns.tolist()

# Remove whitespace around column names
columns = [col.strip() for col in columns]

# Remove Label
feature_names = [col for col in columns if col != "Label"]

print("Number of features:", len(feature_names))
print("\nFeature names:")
for i, feature in enumerate(feature_names):
    print(f"{i}: {feature}")
```

```python
import json

feature_names_path = Path(r"C:\SentinelXAI\data\processed\feature_names.json")

with open(feature_names_path, "w") as f:
    json.dump(feature_names, f, indent=2)

print(f"Saved to: {feature_names_path}")
```
